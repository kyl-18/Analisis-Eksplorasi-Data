Mencari data sebanyak 100 sampel

```python
# Import library yang dibutuhkan
import numpy as np
import pandas as pd
import scipy.stats as stats

# Mengatur random seed agar data konsisten dan tidak berubah-ubah setiap kali dijalankan
np.random.seed(42)

# Membangkitkan data sebanyak 100 sampel (misalnya berdistribusi normal dengan mean=50 dan std=10)
data = np.random.normal(loc=50, scale=10, size=100)

print("Data sampel (10 data pertama):")
print(data)
print(f"Jumlah total sampel: {len(data)}")
```

Menghitung nilai estimasi titik dan interval menggunakan distribusi Z dengan tingkat kesalahan 5%.

```python
# Menghitung statistik deskriptif dasar
n = len(data)
titik_tengah_mean = np.mean(data)
std_dev = np.std(data, ddof=1) # Standar deviasi sampel

# Tingkat kesalahan 5% berarti tingkat kepercayaan (confidence level) = 95% (0.95)
alpha = 0.05
z_score = stats.norm.ppf(1 - alpha / 2) # Nilai Z untuk alpha 5% (~1.96)

# Menghitung margin of error dan interval
margin_error = z_score * (std_dev / np.sqrt(n))
lower_bound = titik_tengah_mean - margin_error
upper_bound = titik_tengah_mean + margin_error

print("HASIL ESTIMASI TITIK DAN INTERVAL (DISTRIBUSI Z)")
print(f"a. Estimasi Titik (Rata-rata/Mean): {titik_tengah_mean:.4f}")
print(f"   Nilai Z (alpha 5%): {z_score:.4f}")
print(f"   Margin of Error: {margin_error:.4f}")
print(f"   Interval Kepercayaan 95%: ({lower_bound:.4f}, {upper_bound:.4f})")
```

Melakukan resampling menggunakan metode jackknife dan bootstrapping, hitung nilai rata-rata dari hasil resamplingnya. untuk bootstrapping lakukan iterasi sebanyak 1000.

```python
# Proses resampling Jackknife
jackknife_means = []

for i in range(n):
    # Menghapus satu data pada indeks ke-i
    subset = np.delete(data, i)
    jackknife_means.append(np.mean(subset))

rata_rata_jackknife = np.mean(jackknife_means)

print("HASIL RESAMPLING JACKKNIFE")
print(f"Jumlah iterasi Jackknife: {len(jackknife_means)}")
print(f"Rata-rata dari hasil resampling Jackknife: {rata_rata_jackknife:.4f}")
```

```python
# Proses resampling Bootstrapping sebanyak 1000 iterasi
n_iterations = 1000
bootstrap_means = []

for _ in range(n_iterations):
    # Mengambil sampel dengan pengembalian berukuran sama dengan data asli (n=100)
    bootstrap_sample = np.random.choice(data, size=n, replace=True)
    bootstrap_means.append(np.mean(bootstrap_sample))

rata_rata_bootstrap = np.mean(bootstrap_means)

print("HASIL RESAMPLING BOOTSTRAPPING")
print(f"Jumlah iterasi Bootstrapping: {n_iterations}")
print(f"Rata-rata dari hasil resampling Bootstrapping: {rata_rata_bootstrap:.4f}")
```
